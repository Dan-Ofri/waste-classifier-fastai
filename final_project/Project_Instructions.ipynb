### 🔍 Overview
Design, train, and analyze **convolutional neural networks (CNNs)** for an image classification problem of your choice (the same one selected in Exercise 1). You previously built a dataset (~1,000 images) and trained a classifier using **fastai**. In this project, you will implement and train your own network in **PyTorch** (without fastai).

> **Focus is not only on accuracy.** Emphasize architectural choices, optimization strategies, hyperparameter tuning, overfitting vs generalization, and explaining *why* things worked or didn't.

### 🎯 Learning Goals
- Design and train CNN architectures for image classification  
- Experiment with optimization algorithms and hyperparameters  
- Apply batch normalization and regularization techniques  
- Use transfer learning effectively  
- Work with pretrained networks (e.g., ResNet50)  
- Analyze and communicate experimental results clearly  

---

- **Part 1 — Training a CNN from Scratch**

  Adapt ResNet50, replace the final layer, experiment with freezing/unfreezing and fine-tuning.

  <summary><strong>Part 2 — Transfer Learning</strong></summary>

  ### Step 4 — Pretraining on an External Dataset
  
  > **⚠️ IMPORTANT CLARIFICATION:**  
  > In this part, you will **design and train a network from scratch** on the external dataset. This means:
  > - **Architectural design:** Choose/design an architecture (can be similar to Part 1, but doesn't have to be!)
  > - **Complete training:** Train from random initialization on the external dataset
  > - **No weight reuse:** Do NOT load or reuse weights from Part 1
  > 
  > **Why "from scratch"?** Because different datasets may benefit from different architectures!
  > - Part 1: ~1,000 images, specific domain (e.g., waste)
  > - Part 2: Potentially 60,000 images (CIFAR-10), different domain
  > 
  > **This is your opportunity to discover** that what worked in Part 1 might not be optimal here!

  1. Choose an external image dataset (e.g., CIFAR-10 or another public dataset).  
  2. **Design a CNN architecture** for this dataset. Consider:
     - Should you use the same architecture as Part 1, or try something different?
     - Does the image size (e.g., 32×32 for CIFAR-10 vs larger images) affect your design?
     - Does having more data allow for a deeper network?
  3. **Train from scratch** on this dataset (or a subset if it's too large).
  4. Save the trained weights.  

  **Key insights to explore:**
  - The external dataset (e.g., CIFAR-10) is likely larger and has a different nature than your dataset
  - What worked well in Part 1 may not be optimal here—experiment and learn from the differences!
  - Compare what gives good results in Part 1 vs Part 2
  - Document how the different dataset characteristics affect your design choices
  - If results are poor, consider redesigning the architecture!

  *Explain why this dataset is relevant and what features it may teach the model.*

  ### Step 5 — Fine-Tuning on Your Dataset
  1. Load the pretrained weights from Step 4.  
  2. Fine-tune the network using your dataset.  
  3. Compare results with training from scratch (Part 1).  

  Analyze:
  - Speed of convergence  
  - Final accuracy  
  - Generalization  
  - How did pre-training on a different dataset affect performance?

  **Notes:**
  1. You may need to adjust image sizes/shapes to match your network's expected input.  
  2. Adjust the final classification layer to the number of classes in the target dataset.
  3. Don't assume that what worked best in Part 1 will be best here—the larger, different dataset may reveal new insights!
